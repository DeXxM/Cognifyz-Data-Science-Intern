```python
import pandas as pd
import matplotlib.pyplot as plt
a=pd.read_csv("Dataset.csv")
print(a.shape)
```

Task 3: Data Visualization

```python
plt.hist(a["Aggregate rating"])
plt.xlabel("Aggregate rating")
plt.ylabel("No. of Restaurants")
plt.title("Histogram of Aggregate rating")
plt.show()
```

```python
avg=a.groupby("Cuisines")["Aggregate rating"].mean().head(10)
print(avg)
```

```python
plt.bar(avg.index, avg.values)
plt.xlabel("Cuisines")
plt.ylabel("Aggregate rating")
plt.title("Bar plot of Aggregate rating vs Cuisines")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.scatter(a["Votes"], a["Aggregate rating"])
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")
plt.title("Votes vs Aggregate Rating")
plt.show()
```
