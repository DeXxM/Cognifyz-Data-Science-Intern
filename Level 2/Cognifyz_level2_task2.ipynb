```python
import pandas as pd
a=pd.read_csv("Dataset.csv")
print(a.shape)
```

Task 2: Price Range Analysis

```python
print(a["Price range"].value_counts())
h=a["Price range"].value_counts().idxmax()
print("Most common prize range-",h,":",a["Price range"].value_counts().max())
```

```python
print(a.groupby(["Price range"])["Aggregate rating"].mean())
```

```python
b=a.groupby(["Price range"])["Aggregate rating"].mean().max()
c=a.groupby(["Price range"])["Aggregate rating"].mean().idxmax()
print("highest average rating- ",b)
print(a[a["Price range"]==c]["Rating color"].value_counts())
```
