Task 1: Table Booking and Online Delivery

```python
print(a["Has Table booking"].value_counts())
print(a["Is delivering now"].value_counts())
```

```python
yes1=a["Has Table booking"].value_counts()["Yes"]
yes2=a["Is delivering now"].value_counts()["Yes"]
t=len(a)

p1=(yes1/t)*100
p2=(yes2/t)*100

print("Has Table booking = ",p1,"%")
print("Is delivering now = ",p2,"%")
```
